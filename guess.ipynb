```python
import random

def number_game():
    number_to_guess = random.randint(1, 100)
    attempts = 0
    max = 10

    print("Number Guessing Game")
    print("number between 1 to 100.")
    print(f"You have {max} attempts to guess.")

    while attempts < max:
        try:
            guess = int(input(f"Attempt {attempts + 1}: Enter your guess: "))
            attempts += 1

            if guess < number_to_guess:
                print("Too low!")
            elif guess > number_to_guess:
                print("Too high!")
            else:
                print(f" Waww You guessed the number in {attempts} attempts.")
                break
        except ValueError:
            print("Invalid input")

    if attempts == max and guess != number_to_guess:
        print(f" You've run out of attempts! The number was {number_to_guess}.")

number_game()
```
